## Task 3: A Veracity Check

```python
print("Missing Values: ")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDataset Shape Before Cleaning:")
print(df.shape)

df_clean = df.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(df_clean.shape)

#count unique rows with at least one data-quality issue
problem_rows = (
    df.isnull().any(axis=1) | df.duplicated()
).sum()

problem_rate = (problem_rows / len(df)) * 100

print("\nTotal affected rows:", problem_rows)
print("\nCombined problem rate:", round(problem_rate, 2), "%")
```


### Task 3: Results and Interpretation

- **Columns with missing values:** 
  - monthly_income_etb: 13 and
  -  has_collateral: 6
- **Number of exact duplicate rows:** 5
- **Total missing values:** 19
- **Dataset before cleaning:** 305 rows, 8 columns.
- **Dataset after removing duplicates:** 300 rows, 8 columns.
- **Combined problem rate:** 7.54%

**Interpretation:**

The dataset contains missing values in the monthly_income_etb and has_collateral columns. It also contains five exact duplicate rows. These issues may affect the reliability of analysis, so they should be considered before using the dataset for decision-making.

## Task 5 (stretch) - Conditional Probability, Revisited

```python
#Overall probability of default
p_default = (df["defaulted"] == "Yes").mean()
print("P(Default):", p_default)

#Probability of default based on collateral or no collateral
rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)

print("\nDefault Rate by Collateral:")
print(rate_by_collateral)
```


### Task 5: Results and Interpretation

- **P(Default):** 0.193443 (19.34%)
- **P(Default | Has Collateral):** 0.120805 (12.08%)
- **P(Default | No Collateral):** 0.260000 (26.00%)

**Interpretation:**

The results show that applicants without collateral have a higher default rate (26%) than applicants with collateral (12.08%)

The difference between the two conditional default rates is approximately 13.92 percentage points. 
Applicants without collateral default more frequently
